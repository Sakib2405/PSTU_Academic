```python
import random
import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return -(x - 3) ** 2 + 9

def hill_climbing(max_iterations=100, step_size=0.1):
    current = random.uniform(-10, 10)
    history = [current]

    for _ in range(max_iterations):
        neighbor = current + random.uniform(-step_size, step_size)
        if f(neighbor) > f(current):
            current = neighbor
            history.append(current)

    return current, f(current), history

solution, value, history = hill_climbing()

x = np.linspace(-10, 10, 400)
y = f(x)

plt.plot(x, y, label="f(x)", color="blue")
plt.title("Hill Climbing Visualization")
plt.xlabel("x")
plt.ylabel("f(x)")

hx = history
hy = [f(h) for h in history]
plt.plot(hx, hy, "ro-", label="Search Path")

plt.legend()
plt.grid(True)
plt.show()

print("Best solution found:", solution)
print("Function value:", value)

```

f(x) → আমাদের objective function।
এখানে f(x) = -(x - 3)^2 + 9।

সর্বোচ্চ মান (maximum) পাওয়া যায় x=3 এ।

সেই জায়গায় f(3) = 9।

current → শুরুতে -10 থেকে 10 এর মধ্যে এলোমেলো একটা মান নেওয়া হয়।

Neighbor তৈরি করা → current এর কাছাকাছি (step_size = 0.1 এর মধ্যে) একটা নতুন মান neighbor বের করা হয়।

Comparison করা →

যদি neighbor এর মান current এর চাইতে বেশি হয়, তাহলে আমরা neighbor কে নতুন solution হিসেবে গ্রহণ করি।

না হলে কিছুই করি না (মানে stuck হয়ে যেতে পারি)।

Loop → এই কাজটা অনেকবার (max_iterations=1000) করা হয়।

শেষে → algorithm একটা ভালো solution রিটার্ন করে।

সাধারণত global maximum (x=3, value=9) এর কাছাকাছি চলে আসে।

তবে drawback হলো → Hill Climbing local maximum এ আটকে যেতে পারে।
