```python
import random

# Example graph (adjacency list with heuristic values)
graph = {
    'A': [('B', 7), ('C', 6)],
    'B': [('D', 4), ('E', 5)],
    'C': [('F', 3), ('G', 4)],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

# Heuristic values (goal is node 'G')
heuristic = {
    'A': 8,
    'B': 7,
    'C': 6,
    'D': 4,
    'E': 5,
    'F': 3,
    'G': 0   # Goal node
}

def local_beam_search(start, goal, k=2):
    # Start with k random states (here, just expand from start)
    states = [start]
    print(f"Initial state: {states}")

    while True:
        # If any state is the goal
        if goal in states:
            print(f"Goal {goal} found!")
            return goal

        # Generate successors
        successors = []
        for state in states:
            for neighbor, _ in graph[state]:
                successors.append(neighbor)

        if not successors:
            print("No more successors, search failed.")
            return None

        # Sort successors by heuristic value
        successors.sort(key=lambda x: heuristic[x])

        # Keep best k successors
        states = successors[:k]
        print(f"Selected states: {states}")

# Run Local Beam Search
local_beam_search('A', 'G', k=2)
```
