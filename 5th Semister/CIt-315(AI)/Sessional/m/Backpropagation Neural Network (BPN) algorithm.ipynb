```python
import math

def sigmoid(x): return 1.0/(1.0+math.exp(-x))
def dsig(x): return x*(1-x)

i1,i2=0.05,0.10
t1,t2=0.01,0.99
w1,w2,w3,w4=0.15,0.20,0.25,0.30
w5,w6,w7,w8=0.40,0.45,0.50,0.55
b_h,b_o=0.35,0.60
lr=0.5

# forward
h1_in=i1*w1+i2*w3+b_h
h2_in=i1*w2+i2*w4+b_h
h1=sigmoid(h1_in)
h2=sigmoid(h2_in)
o1_in=h1*w5+h2*w7+b_o
o2_in=h1*w6+h2*w8+b_o
o1=sigmoid(o1_in)
o2=sigmoid(o2_in)

# error
E=0.5*((t1-o1)**2+(t2-o2)**2)

# output layer deltas
delta_o1=(o1-t1)*dsig(o1)
delta_o2=(o2-t2)*dsig(o2)

# hidden deltas
delta_h1=(delta_o1*w5+delta_o2*w6)*dsig(h1)
delta_h2=(delta_o1*w7+delta_o2*w8)*dsig(h2)

# weight updates
w5 -= lr*delta_o1*h1
w6 -= lr*delta_o2*h1
w7 -= lr*delta_o1*h2
w8 -= lr*delta_o2*h2

w1 -= lr*delta_h1*i1
w3 -= lr*delta_h1*i2
w2 -= lr*delta_h2*i1
w4 -= lr*delta_h2*i2

b_o -= lr*(delta_o1+delta_o2)
b_h -= lr*(delta_h1+delta_h2)

print(round(w1,9),round(w2,9),round(w3,9),round(w4,9),round(w5,9),round(w6,9),round(w7,9),round(w8,9))
print(round(b_h,9),round(b_o,9))
print("error",E)
```
