```python
from queue import PriorityQueue
graph={'A':{'B':2,'C':4},'B':{'C':1,'D':7},'C':{'E':3},'D':{'F':1},'E':{'D':2,'F':5},'F':{}}
h={'A':10,'B':8,'C':5,'D':7,'E':3,'F':0}
def astar(s,g):
    q=PriorityQueue()
    q.put((h[s],[s]))
    while not q.empty():
        c,p=q.get()
        n=p[-1]
        if n==g:return p
        for nb in graph[n]:
            np=p+[nb]
            g1=sum(graph[p[i]][p[i+1]] for i in range(len(p)-1))
            f=g1+graph[n][nb]+h[nb]
            q.put((f,np))
print(astar('A','F'))
```
