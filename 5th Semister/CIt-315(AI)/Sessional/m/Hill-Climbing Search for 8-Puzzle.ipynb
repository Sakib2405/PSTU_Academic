Initial:

3 8 5

2 1 _

6 7 4



Goal:

1 2 3

4 5 6

7 8 _


```python
import random

goal=[[1,2,3],[4,5,6],[7,8,0]]
init=[[3,8,5],[2,1,0],[6,7,4]]

def h(s):
    d=0
    for i in range(3):
        for j in range(3):
            v=s[i][j]
            if v!=0:
                x=(v-1)//3
                y=(v-1)%3
                d+=abs(x-i)+abs(y-j)
    return d

def moves(s):
    res=[]
    for i in range(3):
        for j in range(3):
            if s[i][j]==0:
                x,y=i,j
    dirs=[(1,0),(-1,0),(0,1),(0,-1)]
    for dx,dy in dirs:
        nx,ny=x+dx,y+dy
        if 0<=nx<3 and 0<=ny<3:
            ns=[r[:] for r in s]
            ns[x][y],ns[nx][ny]=ns[nx][ny],ns[x][y]
            res.append(ns)
    return res

def hill(s):
    while True:
        cur=h(s)
        neigh=moves(s)
        better=min(neigh,key=h)
        if h(better)>=cur:
            return s
        s=better

res=hill(init)
for r in res:print(r)
```
