```python
import random
def cost(b):
    c=0
    for i in range(8):
        for j in range(i+1,8):
            if b[i]==b[j] or abs(b[i]-b[j])==abs(i-j):c+=1
    return c
def hill():
    b=[random.randint(0,7) for _ in range(8)]
    while True:
        c0=cost(b)
        best=b
        for i in range(8):
            for j in range(8):
                nb=b[:]
                nb[i]=j
                if cost(nb)<cost(best):best=nb
        if cost(best)>=c0:return b
        b=best
r=hill()
print(r)
```
