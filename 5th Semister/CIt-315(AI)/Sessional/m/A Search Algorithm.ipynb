```python
from queue import PriorityQueue

graph = {
    'A': {'B': 2, 'E': 11},
    'B': {'A': 2, 'C': 6, 'D': 9},
    'C': {'B': 6, 'G': 99},
    'D': {'B': 9, 'E': 3, 'G': 1},
    'E': {'A': 11, 'D': 3},
    'G': {}
}
h = {'A':10, 'B':8, 'C':99, 'D':7, 'E':11, 'G':0}

def astar(start, goal):
    pq = PriorityQueue()
    pq.put((h[start], [start]))
    while not pq.empty():
        cost, path = pq.get()
        node = path[-1]
        if node == goal:
            return path
        for neighbor in graph[node]:
            new_path = list(path)
            new_path.append(neighbor)
            g = sum(graph[path[i]][path[i+1]] for i in range(len(path)-1))
            f = g + h[neighbor]
            pq.put((f, new_path))
print(astar('A', 'G'))
```
