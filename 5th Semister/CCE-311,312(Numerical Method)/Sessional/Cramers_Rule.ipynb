```python
import numpy as np

def cramers_rule(A, B):
    det_A = np.linalg.det(A)
    if det_A == 0:
        return "No unique solution (determinant = 0)."
    
    n = A.shape[0]
    solutions = []
    
    for i in range(n):
        Ai = A.copy()
        Ai[:, i] = B  # replace i-th column with B
        det_Ai = np.linalg.det(Ai)
        solutions.append(det_Ai / det_A)
    
    return solutions

# Example: Solve system
# 2x + y - z = 8
# -3x - y + 2z = -11
# -2x + y + 2z = -3

A = np.array([[2, 1, -1],
              [-3, -1, 2],
              [-2, 1, 2]], dtype=float)

B = np.array([8, -11, -3], dtype=float)

solution = cramers_rule(A, B)
print("Solution (x, y, z) =", solution)
```
