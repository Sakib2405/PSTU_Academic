#### Approximate the integral 

I = ∫₀¹ e^(-x²) dx 

using the Composite Trapezoidal Rule with n = 4 and n = 8 subintervals. Compare each approximation with the exact value 

I_exact = (√π / 2) * erf(1).


```python
import numpy as np
import math
import matplotlib.pyplot as plt

# Trapezoidal Rule function
def trapezoidal(f, a, b, n):
    a = float(a); b = float(b); n = int(n)
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    y = f(x)
    return h * (0.5 * y[0] + y[1:-1].sum() + 0.5 * y[-1])

# Function and exact integral
f = lambda x: np.exp(-x**2)
a, b = 0.0, 1.0
exact = (math.sqrt(math.pi)/2.0) * math.erf(1.0)

# Test for n = 4 and n = 8
for n in [4, 8]:
    approx = trapezoidal(f, a, b, n)
    error = abs(exact - approx)
    print(f"n={n}: Approximation = {approx:.8f}, Error = {error:.8e}")

    # Plot trapezoids
    x = np.linspace(a, b, n + 1)
    y = f(x)
    xx = np.linspace(a, b, 500)
    yy = f(xx)

    plt.figure(figsize=(7,4))
    plt.plot(xx, yy, 'b', label='f(x)=e^{-x^2}')
    for i in range(n):
        xs = [x[i], x[i], x[i+1], x[i+1]]
        ys = [0, y[i], y[i+1], 0]
        plt.fill(xs, ys, alpha=0.5, edgecolor='k')
    plt.title(f"Trapezoidal Approximation (n={n})")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.grid(True)
    plt.show()
```
