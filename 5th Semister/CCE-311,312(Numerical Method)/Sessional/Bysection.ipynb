```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

def f(x):
    return x**3 - 2*x - 5

def bisection(a,b,tol):
    xl = a
    xr = b
    i = 1
    while(np.abs(xl-xr) >= tol):
        c = (xl+xr)/2.0
        product = f(xl) * f(c)
        if product > tol:
            xl = c
            i+=1
        else:
            if product < tol:
                xr = c
                i+=1
    return i,c

answer = bisection(-5,5,1e-8)
print("Bisection Method Gives Root At x = ", answer[1])
print("Iteration: ",answer[0])

import inspect

a = inspect.signature(fsolve)
print(a)

#insted of using inspect module we may use shortcut..
'''
shortcut = fsolve(f,[-5,1.5])
print ("Fsolve gives 1st root at x = ",shortcut[0])

'''
        
x = np.linspace(-5,5,100)
plt.plot(x,f(x))
plt.grid()
plt.show()
```
