```python
def f(x):
    return x**3 - x - 2   # Example function

def false_position(a, b, tol=1e-6, max_iter=100):
    if f(a) * f(b) >= 0:
        print("Invalid interval: f(a) and f(b) must have opposite signs")
        return None
    
    print("Iter\t   a\t\t   b\t\t   c\t\t f(c)")
    for i in range(1, max_iter+1):
        c = (a*f(b) - b*f(a)) / (f(b) - f(a))
        print(f"{i}\t{a:.6f}\t{b:.6f}\t{c:.6f}\t{f(c):.6f}")
        
        if abs(f(c)) < tol:
            print(f"\nRoot found at x = {c:.6f} after {i} iterations")
            return c
        
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    

    print("\nMethod did not converge")
    return None

false_position(1, 2)

```
