```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def logistic_regression():
    n = int(input("Enter number of data points: "))
    x = []
    y = []
    for i in range(n):
        xi = float(input(f"Enter feature x[{i}]: "))
        yi = int(input(f"Enter label y[{i}] (0 or 1): "))
        x.append([1, xi])  # adding bias term
        y.append(yi)

    x = np.array(x)
    y = np.array(y)
    lr = float(input("Enter learning rate: "))
    epochs = int(input("Enter number of epochs: "))
    w = np.zeros(x.shape[1])

    for _ in range(epochs):
        z = np.dot(x, w)
        h = sigmoid(z)
        gradient = np.dot(x.T, (y - h))
        w += lr * gradient

    print("Weights:", w)

logistic_regression()
```

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def logistic_regression():
    x = np.array([[1, 0], [1, 1], [1, 2], [1, 3]])
    y = np.array([0, 0, 1, 1])
    lr = 0.1
    epochs = 1000
    w = np.zeros(x.shape[1])

    for _ in range(epochs):
        z = np.dot(x, w)
        h = sigmoid(z)
        gradient = np.dot(x.T, (y - h))
        w += lr * gradient

    print(w)

logistic_regression()
```
