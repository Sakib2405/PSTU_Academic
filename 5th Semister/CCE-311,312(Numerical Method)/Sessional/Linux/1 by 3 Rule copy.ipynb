```python
def simpson_one_third():
    n = int(input("Enter number of intervals (even): "))
    a = float(input("Enter lower limit: "))
    b = float(input("Enter upper limit: "))
    func = input("Enter the function in terms of x: ")

    if n % 2 != 0:
        print("Number of intervals must be even")
        return

    h = (b - a) / n
    result = eval(func.replace('x', str(a))) + eval(func.replace('x', str(b)))
    for i in range(1, n):
        x = a + i * h
        if i % 2 == 0:
            result += 2 * eval(func.replace('x', str(x)))
        else:
            result += 4 * eval(func.replace('x', str(x)))
    result *= h / 3
    print("Integral value:", result)

simpson_one_third()
```
