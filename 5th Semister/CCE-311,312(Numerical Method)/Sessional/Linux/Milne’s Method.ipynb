```python
def milne_method():
    func = input("Enter dy/dx in terms of x and y: ")
    x0 = float(input("Enter initial x value: "))
    y0 = float(input("Enter initial y value: "))
    xn = float(input("Enter x value to find y at: "))
    h = float(input("Enter step size h: "))

    n = int((xn - x0) / h)
    x = [x0 + i*h for i in range(4)]
    y = [y0]
    for i in range(1, 4):
        y.append(y[i-1] + h * eval(func.replace('x', str(x[i-1])).replace('y', str(y[i-1]))))

    for i in range(3, n):
        f1 = eval(func.replace('x', str(x[i-3])).replace('y', str(y[i-3])))
        f2 = eval(func.replace('x', str(x[i-2])).replace('y', str(y[i-2])))
        f3 = eval(func.replace('x', str(x[i-1])).replace('y', str(y[i-1])))
        f4 = eval(func.replace('x', str(x[i])).replace('y', str(y[i])))

        y_pred = y[i-3] + (4*h/3)*(2*f3 - f2 + 2*f1)
        x_next = x[i] + h
        f_pred = eval(func.replace('x', str(x_next)).replace('y', str(y_pred)))
        y_corr = y[i-1] + (h/3)*(f4 + 4*f3 + f_pred)

        x.append(x_next)
        y.append(y_corr)

    print("Value of y at x =", xn, "is", y[-1])

milne_method()
```
