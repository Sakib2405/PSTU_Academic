```python
def simpson_three_eighth():
    n = int(input("Enter number of intervals (multiple of 3): "))
    a = float(input("Enter lower limit: "))
    b = float(input("Enter upper limit: "))
    func = input("Enter the function in terms of x: ")

    if n % 3 != 0:
        print("Number of intervals must be a multiple of 3")
        return

    h = (b - a) / n
    result = eval(func.replace('x', str(a))) + eval(func.replace('x', str(b)))
    for i in range(1, n):
        x = a + i * h
        if i % 3 == 0:
            result += 2 * eval(func.replace('x', str(x)))
        else:
            result += 3 * eval(func.replace('x', str(x)))
    result *= 3 * h / 8
    print("Integral value:", result)

simpson_three_eighth()
```
