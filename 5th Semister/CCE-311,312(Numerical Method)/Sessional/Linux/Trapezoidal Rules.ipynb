```python
def trapezoidal_rule():
    n = int(input("Enter number of intervals: "))
    a = float(input("Enter lower limit: "))
    b = float(input("Enter upper limit: "))
    func = input("Enter the function in terms of x: ")

    h = (b - a) / n
    result = 0.5 * (eval(func.replace('x', str(a))) + eval(func.replace('x', str(b))))
    for i in range(1, n):
        x = a + i * h
        result += eval(func.replace('x', str(x)))
    result *= h
    print("Integral value:", result)

trapezoidal_rule()
```
