```python
def picard_method():
    func = input("Enter f(x, y) in terms of x and y: ")
    x0 = float(input("Enter initial x value: "))
    y0 = float(input("Enter initial y value: "))
    xn = float(input("Enter x value to find y at: "))
    n = int(input("Enter number of iterations: "))
    h = float(input("Enter step size h: "))

    x = x0
    y = y0

    for i in range(n):
        y_new = y0
        for j in range(i+1):
            y_new += h * eval(func.replace('x', str(x0 + j*h)).replace('y', str(y0)))
        y = y_new
    print("Approximate value of y at x =", xn, "is", y)

picard_method()
```
