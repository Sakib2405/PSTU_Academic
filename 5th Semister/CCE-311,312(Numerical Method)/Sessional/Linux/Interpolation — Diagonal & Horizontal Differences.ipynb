```python
def forward_difference_interpolation():
    n = int(input("Enter number of data points: "))
    x = []
    y = []
    for i in range(n):
        xi = float(input(f"Enter x[{i}]: "))
        yi = float(input(f"Enter y[{i}]: "))
        x.append(xi)
        y.append(yi)

    diff_table = [y.copy()]
    for i in range(1, n):
        diff = []
        for j in range(n - i):
            diff.append(diff_table[i - 1][j + 1] - diff_table[i - 1][j])
        diff_table.append(diff)

    print("\nForward Difference Table:")
    for i in range(n):
        row = []
        for j in range(n - i):
            row.append(diff_table[i][j])
        print(row)

    xi = float(input("Enter value to interpolate: "))
    h = x[1] - x[0]
    u = (xi - x[0]) / h
    result = y[0]
    u_term = 1
    fact = 1
    for i in range(1, n):
        u_term *= (u - i + 1)
        fact *= i
        result += (u_term / fact) * diff_table[i][0]

    print(f"Interpolated value at {xi} is {result}")

forward_difference_interpolation()
```
