```python
def runge_kutta():
    func = input("Enter dy/dx in terms of x and y: ")
    x0 = float(input("Enter initial x value: "))
    y0 = float(input("Enter initial y value: "))
    xn = float(input("Enter x value to find y at: "))
    h = float(input("Enter step size h: "))

    n = int((xn - x0)/h)
    x = x0
    y = y0

    for i in range(n):
        k1 = h * eval(func.replace('x', str(x)).replace('y', str(y)))
        k2 = h * eval(func.replace('x', str(x + h/2)).replace('y', str(y + k1/2)))
        k3 = h * eval(func.replace('x', str(x + h/2)).replace('y', str(y + k2/2)))
        k4 = h * eval(func.replace('x', str(x + h)).replace('y', str(y + k3)))
        y += (k1 + 2*k2 + 2*k3 + k4)/6
        x += h

    print("Value of y at x =", xn, "is", y)

runge_kutta()
```
