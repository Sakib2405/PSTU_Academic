```python
import math

func_str = input("Enter the function in terms of x: ")

def g(x):
    return eval(func_str)

x = float(input("Enter initial guess: "))
n = int(input("Enter number of iterations: "))

for i in range(n):
    x = g(x)
    print(x)
```
