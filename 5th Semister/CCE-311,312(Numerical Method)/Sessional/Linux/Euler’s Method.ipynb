```python
def euler_method():
    func = input("Enter the function dy/dx in terms of x and y: ")
    x0 = float(input("Enter initial x value: "))
    y0 = float(input("Enter initial y value: "))
    xn = float(input("Enter value of x to find y at: "))
    h = float(input("Enter step size h: "))

    n = int((xn - x0) / h)
    x = x0
    y = y0

    for i in range(n):
        y += h * eval(func.replace('x', str(x)).replace('y', str(y)))
        x += h

    print(f"Value of y at x = {xn} is {y}")

euler_method()
```
