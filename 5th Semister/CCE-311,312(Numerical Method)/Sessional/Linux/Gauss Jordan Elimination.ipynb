```python
def gauss_jordan():
    A = []
    for i in range(3):
        row = list(map(float, input(f"Enter row {i+1} of coefficients and constant: ").split()))
        A.append(row)

    for i in range(3):
        A[i] = [x / A[i][i] for x in A[i]]
        for j in range(3):
            if i != j:
                ratio = A[j][i]
                A[j] = [A[j][k] - ratio * A[i][k] for k in range(4)]

    X = [A[i][3] for i in range(3)]
    print("Solution:", X)

gauss_jordan()
```
