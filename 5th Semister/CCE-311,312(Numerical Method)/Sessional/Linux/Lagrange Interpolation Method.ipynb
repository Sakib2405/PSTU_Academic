```python
def lagrange_interpolation():
    n = int(input("Enter number of data points: "))
    x = []
    y = []
    for i in range(n):
        xi = float(input(f"Enter x[{i}]: "))
        yi = float(input(f"Enter y[{i}]: "))
        x.append(xi)
        y.append(yi)

    xi = float(input("Enter value to interpolate: "))
    result = 0
    for i in range(n):
        term = y[i]
        for j in range(n):
            if j != i:
                term *= (xi - x[j]) / (x[i] - x[j])
        result += term

    print(f"Interpolated value at {xi} is {result}")

lagrange_interpolation()
```
