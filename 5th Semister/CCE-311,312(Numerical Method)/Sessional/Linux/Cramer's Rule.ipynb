```python
import numpy as np

A = []
for i in range(3):
    row = list(map(float, input(f"Enter row {i+1} of coefficients: ").split()))
    A.append(row)
B = list(map(float, input("Enter constants: ").split()))

A = np.array(A)
B = np.array(B)
det_A = np.linalg.det(A)

if det_A != 0:
    X = []
    for i in range(3):
        Ai = A.copy()
        Ai[:, i] = B
        X.append(np.linalg.det(Ai)/det_A)
    print("Solution:", X)
else:
    print("No unique solution")
```

```python
import numpy as np

A = np.array([[2, -1, 3],
              [1, 0, -2],
              [3, 2, -1]])
B = np.array([5, -1, 4])
det_A = np.linalg.det(A)

if det_A != 0:
    X = []
    for i in range(3):
        Ai = A.copy()
        Ai[:, i] = B
        X.append(np.linalg.det(Ai)/det_A)
    print(X)
else:
    print("No unique solution")
```
