```python
A = []
for i in range(3):
    row = list(map(float, input(f"Enter row {i+1} of coefficients and constant: ").split()))
    A.append(row)

for i in range(3):
    for j in range(i+1, 3):
        ratio = A[j][i]/A[i][i]
        for k in range(4):
            A[j][k] -= ratio * A[i][k]

X = [0, 0, 0]
X[2] = A[2][3]/A[2][2]
X[1] = (A[1][3]-A[1][2]*X[2])/A[1][1]
X[0] = (A[0][3]-A[0][2]*X[2]-A[0][1]*X[1])/A[0][0]

print("Solution:", X)
```
